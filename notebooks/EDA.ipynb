```python
# Import libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

print("Libraries imported successfully!")
```

```python
# Create output directories

os.makedirs("../outputs/graphs", exist_ok=True)
os.makedirs("../outputs/results", exist_ok=True)
os.makedirs("../outputs/cleaned_data", exist_ok=True)

print("Output folders are ready!")
```

```python
# Load datasets

obesity_df = pd.read_csv("../data/obesity_dataset.csv")
hospital_df = pd.read_csv("../data/Hospital_Patient.csv")
healthcare_df = pd.read_csv("../data/healthcare_dataset.csv")

print("All datasets loaded successfully!")
```

```python
# Create copies for further analysis

obesity_clean = obesity_df.copy()
hospital_clean = hospital_df.copy()
healthcare_clean = healthcare_df.copy()

print("Clean copies created successfully!")
```

```python
# Remove duplicate records

obesity_clean = obesity_clean.drop_duplicates()
hospital_clean = hospital_clean.drop_duplicates()
healthcare_clean = healthcare_clean.drop_duplicates()

print("Duplicates removed.")
print("Obesity:", obesity_clean.shape)
print("Hospital:", hospital_clean.shape)
print("Healthcare:", healthcare_clean.shape)
```

```python
# Save cleaned datasets

obesity_clean.to_csv("../outputs/cleaned_data/obesity_cleaned.csv", index=False)
hospital_clean.to_csv("../outputs/cleaned_data/hospital_cleaned.csv", index=False)
healthcare_clean.to_csv("../outputs/cleaned_data/healthcare_cleaned.csv", index=False)

print("Cleaned datasets saved successfully!")
```

```python
# Convert admission and discharge dates

healthcare_clean["Date of Admission"] = pd.to_datetime(
    healthcare_clean["Date of Admission"],
    errors="coerce"
)

healthcare_clean["Discharge Date"] = pd.to_datetime(
    healthcare_clean["Discharge Date"],
    errors="coerce"
)

print("Date conversion completed.")
```

```python
# Calculate length of stay in days

healthcare_clean["Length of Stay"] = (
    healthcare_clean["Discharge Date"] -
    healthcare_clean["Date of Admission"]
).dt.days

display(
    healthcare_clean[
        ["Date of Admission", "Discharge Date", "Length of Stay"]
    ].head()
)
```

```python
# Length of stay distribution

plt.figure(figsize=(8, 5))

sns.histplot(
    data=healthcare_clean,
    x="Length of Stay",
    bins=20,
    kde=True
)

plt.title("Distribution of Hospital Length of Stay")
plt.xlabel("Length of Stay (Days)")
plt.ylabel("Number of Patients")
plt.tight_layout()

plt.savefig("../outputs/graphs/healthcare_length_of_stay.png")
plt.show()
```

```python
# Correlation matrix for obesity dataset

obesity_numeric = obesity_clean.select_dtypes(include=np.number)

plt.figure(figsize=(12, 8))

sns.heatmap(
    obesity_numeric.corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Matrix - Obesity Dataset")
plt.tight_layout()

plt.savefig("../outputs/graphs/obesity_correlation_heatmap.png")
plt.show()
```
